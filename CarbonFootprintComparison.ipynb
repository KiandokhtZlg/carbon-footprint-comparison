Carbon Footprint Comparison: Solar vs. Coal
This project estimates total carbon emission per kwh for solar panels and compares it to coal-based power.


Step 1: Calculate energy production of solar panels:

Assumprtions:
  - Irradiation = 2000 kWh/m²/year
  - Efficiency = 14%  #Mono-crystalline silicon type
  - Performance Ratio = 0.80
  - Lifetime = 30 years

```python
#Variables
irradiation = 2000  # in kWh/m²/year
efficiency = 0.14   # 14%
performance_ratio = 0.80
lifetime_years = 30
carbon_footprint_solar = 40 # g CO2/kWh
carbon_footprint_coal = 1000 # g CO2/kWh
```

```python
electricity = irradiation * efficiency * performance_ratio * lifetime_years
print("Electricity produced over 30 years = ",electricity,"kWh/m²")
```

```python
solar_emission = electricity * carbon_footprint_solar / 1000 #in Kg
coal_emissions = electricity * carbon_footprint_coal / 1000  # in kg
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(9, 5))

sources = ['Solar PV', 'Coal']  # Names shown on the x-axis
emissions = [solar_emission, coal_emission]  # Heights of the bars
colors = ['blue', 'gray']
plt.bar(sources, emissions, color=colors)



plt.ylabel('kg CO2eq')
plt.title('30-Year Carbon Emissions: Solar PV vs. Coal', fontsize=14, weight='bold')
plt.ylabel('kg CO₂ per m²', fontsize=12)
plt.grid(axis='y', linestyle='--')
plt.show()
plt.savefig('result.png')
```

Solar energy produces ~25 times less carbon emissions than coal energy for the same electricity output over a 30-year period.
